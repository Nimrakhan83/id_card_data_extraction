```python
import cv2
import pytesseract
from nltk import ne_chunk, pos_tag, word_tokenize
from nltk.tree import Tree
from nltk.chunk import conlltags2tree, tree2conlltags

# Function to perform OCR using Tesseract
def perform_ocr(image_path):
    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    text = pytesseract.image_to_string(gray, lang='eng')  # Perform OCR
    
    return text

# Function to extract named entities using NLTK
def extract_entities(text):
    entities = []
    words = word_tokenize(text)
    tags = pos_tag(words)
    tree = ne_chunk(tags)
    iob_tagged = tree2conlltags(tree)
    
    for word, tag, iob_tag in iob_tagged:
        if iob_tag != 'O':
            entities.append((word, tag, iob_tag))
    
    return entities

# Example usage
if __name__ == '__main__':
    # Example image path
    image_path = '/kaggle/input/driving-lisense/Driving Lisense.PNG'
    
    # Perform OCR to extract text from image
    extracted_text = perform_ocr(image_path)
    print("Extracted Text:")
    print(extracted_text)
    print()
    
    # Extract named entities (NER)
    named_entities = extract_entities(extracted_text)
    print("Named Entities:")
    for entity in named_entities:
        print(entity)
```
